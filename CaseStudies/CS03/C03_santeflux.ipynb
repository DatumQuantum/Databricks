```python
patients_filtered_df=patients.filter(col("region")==region)
display(patients_filtered_df)
```

```python
patients_filtered_df=patients.withColumn("Age Band",age_band(col("date_of_birth")))
display(patients_filtered_df)
```

```python
def risk_score(dob_col,diag_col,amount_col):
    return when(age_band(dob_col)=="65 and over",2).otherwise(0)
    when(diag_col.startswith("E11") | diag_col.startswith("I10"),1).otherwise(0)
    when(cost_flag(amount_col)=="High",1).otherwise(0)
)
```
