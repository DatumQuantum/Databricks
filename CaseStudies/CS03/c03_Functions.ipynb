```python
from pyspark.sql.functions import *
```

```python
def age_band(dob_col):
    years=floor(datediff(current_date(),dob_col)/365.25)

    return (when(years<18, "under 18")
            .when((years<40) , "18-39")
            .when((years<65) , "40-64")
            .when((years>=65) , "65 and over")
            .otherwise("Invalid")
    )
   
```

```python
def cost_flag(amount_col):
    return when(amount_col>=50,"high").otherwise("Normal")
```
